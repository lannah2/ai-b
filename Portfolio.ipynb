# History in Motion: Reimagining Football Heritage
## Reconstructing Memory Through Human-AI Collaboration

### Introduction of Selected Object 

As the name suggests, Free Kicks at Football is a culturally rich and engaging account of football in Glasgow which resides in the University of Glasgow’s Archives of Special Collections. Robert Maclean (2020) describes the rare piece of memorabilia as “an interesting insight into both the progress of football as a popular sport and the way of life in Victorian Glasgow”. The pamphlet was published in 1882 and spans 19 pages, containing two separate parts. There is very little information available regarding the authors of the artefact, due to their decisions to conceal their identities with alternate titles. The author of Part 1, adopting the presumed pseudonym of ‘Benedict’, composed 4 works: A “Lay-On” the Leather, England v. Scotland, 1882, A Reminiscence of Hampden Park, Our Goal-Keeper Unjerseyed, and A La Mode. Part 2 – written by Saunders Wylie – consists of two works: A Monday Mornin’s Crack and The Dribbler’s Song. Both parts are illustrated by ‘Jingo’ with “satirical and humorous lithographs”.

Whilst there is incredibly little available online data concerning the writers behind the work, or the overall pamphlet for that matter, the content has been analysed and interpreted for the comprehension of a contemporary audience. Maclean (2020) details four out of the six pieces, and the meanings behind the Old English diction. This, however, leaves two pieces uninterpreted: A “Lay-On” The Leather” and Our Goalkeeper Unjerseyed. This portfolio focuses predominantly on Benedict’s A “Lay-On” the Leather, a poem which combines two genres of storytelling: parody and ballad. The piece was constructed as an imitation of the renowned work of Alfred Tennyson, The Brook (1855), and portrays all the components and characteristics of a typical historical football match through the lens of the football itself. Because information is not readily available in the public domain, Artificial Intelligence was utilised to interpret the piece. Below is a compact example of a manner in which AI can augment human interpretation, as will be explored throughout the rest of this portfolio.

### Objectives

In an era of rapidly advancing technology, a predominant form of immersion stems from digital media. As Lin et al. (2022) note, “short-form content would capture a generation accustomed to getting news and entertainment on demand.” Therefore, the primary objective behind this project is to immerse an audience in the content within the pamphlet in a manner to which they are accustomed in contemporary society. The project consists of a narration of the poem A “Lay-On” the Leather, accompanied by a short, movie-style video illustrating the meaning behind the lyrics.

Due to this, the demographic for this project contains no age limit. Young children may find the artefact more comprehensible with the implementation of a visual aid to provide meaning to the words. Young adults, due to decreasing attention spans for written literature in the digital age, may obtain a deeper appreciation of the object through visual and audible stimuli. This may also be applicable to older generations, who may find a straightforward visual and audible account of the history behind the pamphlet more digestible and significant. Furthermore, the visual aspect is beneficial to those who have sight issues. This project has two primary applications: the first is to disseminate the short film digitally, in order to allow those who are unable to visit the original artefact in person to engage with it, increasing accessibility and cultural reach, and widening the demographic of the object. The second is to position the project on a screen next to the location of the object, in order to augment the learning experience by providing a supplementary method of absorption and comprehension.

The new perspective offered by the audible and visual aid is the cultural and societal impact and identity of the sport, which has remained a constant across centuries. Since its inception, football has maintained its status as a global phenomenon, bringing communities together with a shared passion and significantly boosting local economies. Roberts (2015, p.15) states that the sport “functions simultaneously as a consolation for a lack of meaning and as a model and an engine for the desolation of meaning.” This suggests that whilst life can feel without purpose at times, football fills a social gap of belonging and identity. This project aims to display how long football has been a socially significant aspect of the human experience, and this is portrayed particularly in the video through the inclusion of both team and supporter morale alike.

Furthermore, despite the pamphlet containing hand-drawn images, the project demonstrates to audiences in the digital age how artificial intelligence can provoke emotional and immersive experience by reviving and reinvigorating static history. This perspective directly coincides with the frameworks of digital nostalgia and synthetic memory. Digital nostalgia utilises artificial intelligence to generate audiovisual media to reinforce past memory and recreate atmospheres (Shehab, 2025). In this project, digital nostalgia is explored through an audiovisual reimagination of what a typical British football match would have looked like in the 19th century. Synthetic memory consists of using AI to generate memories which supersede lived experience. In this project, synthetic memory assumes the role of constructing what may have been the reality of the poem’s historical context.

A third framework utilised in this project is AI hallucination. This section of the project intends to educate audiences on artificial intelligence’s ability to generate ‘false history’ in a believable and convincing manner. This is illustrated within a separate AI-generated video which was trained on false poetry in the style of the literature residing in the pamphlet. Furthermore, this fictional video, which centres around the increasingly popular women’s football, provides a further level of entertainment to the original object by depicting a light-hearted and engaging ‘what could have been’ trope regarding the history of the sport, taking place during the ban imposed by The FA in 1921, which prevented women from playing for 50 years.


### Methodology

Upon visiting the Special Collections, a plethora of photos were collected from several different angles of each page of the pamphlet, in order to ensure historical accuracy whilst endeavouring to engage a general audience with cultural heritage. During my initial project proposal, I was asked by a peer, “How will you balance creative storytelling with historical accuracy when adapting the manuscript?” Due to this project intending to immerse an audience in the content of the original object, it was pertinent that the content of the selected poem was maintained. As mentioned previously, there is little publicly available information regarding the artefact and, therefore, this is where the human-AI collaboration process commenced. I utilised Gemini, Google’s generative AI assistant, to interpret the poem’s Victorian diction in order to ensure that the visual aspect of the project was compatible with the audible element, and that the tone was accurately depicted. The human input consisted of the original poem, along with the text prompt: “Can you please interpret what this poem means in modern English in simple terms?” Due to the fact that Gemini is a predominantly closed-source large language model, its training data is not available for public access. However, since Google is the creator of Gemini, the model is trained on incredibly large datasets obtained from a multitude of different resources on the internet. Therefore, when inputting the poem, Gemini draws on patterns learned from a variety of online interpretations of 19th-century Victorian English. In doing so, it learns the meanings of words, how they function within a poetic context, and how they can be translated into contemporary English.

## Here is the output from Google Gemini:

This poem, "A 'LAY-ON' THE LEATHER," is about a football (soccer ball), specifically its experience during a match. The "leather" in the title refers to the material the ball is made of.

**Let's break down what each stanza means:**

- *Stanza 1:*
"I'm batter'd, batter'd to and fro, At rest a moment never," - The ball is constantly being kicked and moved around, never still.
"For men may come and men may go, But I go on for ever." - This is a play on a famous line, implying that players change, but the ball (and the game) continues. It's a bit of hyperbole, as balls do wear out, but it captures the continuous nature of the game from the ball's perspective.

- *Stanza 2:*
"I'm bang'd about, now in now out, Now with a dribbler sailing," - The ball is frequently hit, either going into or out of bounds, or being skilfully manoeuvred by a player.
"There's here and there a lusty shout, And here and there a wailing." - This refers to the crowd's reactions: cheers for good plays ("lusty shout") and groans or complaints for bad ones ("wailing").

- *Stanza 3:*
"By thirty kicks I'm hurried down, Or slipp'd beyond the edges," - The ball is kicked many times, sometimes out of bounds.
"By twenty fouls, a broken crown, And half a hundred dodges." - "Broken crown" could metaphorically refer to the ball being heavily impacted or even slightly damaged, or perhaps the 'crown' of a player who gets fouled. The ball is involved in many fouls and dodges.

- *Stanza 4:*
"And here and there away they break, From goal to goal I travel;" - The players make attacking runs, carrying the ball from one end of the field to the other.
"And here and there a score they make, Or grovel in the gravel." - Sometimes a goal is scored, and sometimes players fall or "grovel in the gravel" (or turf, more likely for a modern game) in their efforts.

- *Stanza 5:*
"They slip, they slide, they kick, I'm shied High up among the swallows," - The players' movements are described, and the ball is sometimes kicked high into the air.
"Or make some city rough subside, As round the rope he bellows," - This likely refers to a particularly strong or unruly player ("city rough") who might be shouting intensely during the game. "Round the rope" could refer to the boundary of the playing area or a figurative boundary of his temper.

- *Stanza 6:*
"With many a shove they fume and fret, They make a sudden rally, And many a shy at goal they get By passing to an ally." - Players push, get frustrated, make sudden attacks, and attempt shots at goal by collaborating.

- *Stanza 7:*
"They scream, they roar, they shout " well played." In many a bass and treble," - The crowd and players express their excitement with various vocalizations, from deep voices to high-pitched ones.
"And every nimble run's "hooray'd !" And every supple dribble." - Good plays, like skilful runs and dribbles, are met with cheers.

- *Stanza 8:*
"At last below the bar I go, In spite of all endeavour;" - This is the ultimate moment for the ball: it goes below the crossbar (into the goal), despite the defending team's best efforts.
"For men may come and men may go, But scores run on for ever." - This echoes the first stanza, but with a different emphasis. While players and games come and go, the scores and records of the game are what ultimately endure.

In essence, the poem gives a vivid, personified account of a football match from the perspective of the ball itself. It captures the energy, chaos, excitement, and ultimate purpose of the game.


### Evaluation

I human-checked the interpretation through logical reasoning and rationality and concluded that Gemini’s analysis appeared legitimate. After confirming that its output would be beneficial to my project, I decided to utilise the generated stanza breakdown as the basis of my text-to-video prompts. Upon conducting initial research, I discovered that “text-to-video generators, as well as their text-to-image predecessors, use natural language processing and machine learning algorithms” (Šumilova, 2022). Upon learning this, I endeavoured to ensure that my prompts were as detailed and informative as possible. This is effective regarding my objectives as it guarantees an immersive and informative experience of the original object for a variety of demographics, and allows audiences to observe first-hand how AI can utilise digital nostalgia and synthetic memory to convincingly, and in immense detail, mirror the prolonged significance of football to the human experience. However, I did encounter a degree of setbacks throughout the creative process. For example:

    “Text-to-video conversion is more demanding and complicated than text-to-image generators. The AI generator has to work harder to predict how images shift in time and has to produce numerous images in sequence to capture motion. Another issue is the lack of large datasets with high-quality videos and the issue of modelling data. Also, video frames start to deviate from the prompt if the prompt does not contain sufficient information for subsequent changes.” (Šumilova, 2022)

Due to this, I tested a multitude of different generative AI programmes and ultimately selected the one I believed would allow the most effective level of human-AI collaboration – for example, those which reacted most efficiently to changes in prompts. I then discovered that Google Gemini was in collaboration with Creative Fabrica to create Studio AI, which contains a highly renowned and efficient text-to-video generator. Koigi (2026) states that:

    “The partnership [with Creative Fabrica] will see the digital marketplace integrate Google Cloud’s Gemini Enterprise Agent Platform and multimodal AI models into its creative workflow tools. The company said the agreement is designed to enhance its Studio AI platform, which provides creators with image, video, audio and 3D design tools. Creative Fabrica said the integration of Google Cloud technologies is intended to simplify the content creation process for both hobbyist creators and professional designers by reducing technical complexity. According to the company, the tool combines multiple Gemini prompts and outputs into a single workflow, allowing users to generate 3D previews from text descriptions in real time. The platform currently incorporates several Google Cloud AI technologies, including: Gemini for reasoning and design execution, Veo and Lyria for video and audio generation, and Imagen and Nano Banana for image creation. Creative Fabrica said the partnership supports its broader goal of making advanced design tools more accessible to creators regardless of technical experience.”

These factors led me to conclude that Creative Fabrica’s Studio AI would be the most effective programme for the human-AI collaboration within this project. I input the same initial prompt that I had used in previous AI software options into Studio AI: English football player scoring a goal against the opposing goalkeeper in a 1v1. It is a sunny day. They are playing on a large pitch with slightly dead grass. It is set in the 1880s so the football kits are old-fashioned – they have collars, long sleeves, and long socks with brown football boots. The scorer celebrates with his team after the goal. The clip is in black and white.
