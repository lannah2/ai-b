# History in Motion: Reimagining Football Heritage
## Reconstructing Memory Through Human-AI Collaboration

### Introduction of Selected Object 

As the name suggests, Free Kicks at Football is an engaging and culturally rich account of football in Glasgow which resides in the University of Glasgow’s Archives of Special Collections. Robert Maclean (2020) describes the rare piece of memorabilia as “an interesting insight into both the progress of football as a popular sport and the way of life in Victorian Glasgow”. The pamphlet was published in 1882 and spans 19 pages, containing two separate parts.

There is negligible information available regarding the authors of the artefact, due to their decisions to disguise themselves with alternate titles. The author of part 1, adopting the presumable pseudonym of ‘Benedict’, composed 4 works: A “Lay-On” the Leather, England v. Scotland, 1882, A Reminiscence of Hampden Park, Our Goal-Keeper Unjerseyed, and A La Mode. Part 2 – written by Saunders Wylie – consists of two works: A Monday Mornin’s Crack and The Dribbler’s Song. Both parts are illustrated by ‘Jingo’ with “satirical and humorous lithographs”. Whilst there is incredibly little available online data concerning the writers behind the work, or the overall pamphlet for that matter, the content has been analysed and interpreted for the comprehension of a contemporary audience.

Maclean (2020) details four out of the six pieces, and the meanings between the Old English diction. This however, leaves two pieces uninterpreted: A “Lay-On” The Leather” and Our Goalkeeper Unjerseyed. This portfolio focuses predominantly on Benedict’s A “Lay-On” the Leather, a poem which combines two genres of storytelling: parody and ballad. The piece was constructed as an imitation of the renowned work of Alfred Tennyson, The Brook (1855), and portrays all the components and characteristics of a typical historical football match through the lens of the football itself. Because information is not readily available for public retention, Artificial Intelligence was utilised to interpret the piece. This is a compact example of a manner in which AI can augment human process, as will be explored throughout this portfolio.
